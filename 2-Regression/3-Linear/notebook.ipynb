## Pumpkin Pricing

Load up required libraries and dataset. Convert the data to a dataframe containing a subset of the data: 

- Only get pumpkins priced by the bushel
- Convert the date to a month
- Calculate the price to be an average of high and low prices
- Convert the price to reflect the pricing by bushel quantity

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from datetime import datetime

pumpkins = pd.read_csv('../data/US-pumpkins.csv')

pumpkins.head()
```

```python
pumpkins = pumpkins[pumpkins['Package'].str.contains('bushel', case=True, regex=True)]
#Above we selected desired records.
columns_to_select = ['Package', 'Variety', 'City Name', 'Low Price', 'High Price', 'Date']
pumpkins = pumpkins.loc[:, columns_to_select]

price = (pumpkins['Low Price'] + pumpkins['High Price']) / 2

month = pd.DatetimeIndex(pumpkins['Date']).month
day_of_year = pd.to_datetime(pumpkins['Date']).apply(lambda dt: (dt-datetime(dt.year,1,1)).days)

new_pumpkins = pd.DataFrame(
    {'Month': month, 
     'DayOfYear' : day_of_year, 
     'Variety': pumpkins['Variety'], 
     'City': pumpkins['City Name'], 
     'Package': pumpkins['Package'], 
     'Low Price': pumpkins['Low Price'],
     'High Price': pumpkins['High Price'], 
     'Price': price})

new_pumpkins.loc[new_pumpkins['Package'].str.contains('1 1/9'), 'Price'] = price/1.1
new_pumpkins.loc[new_pumpkins['Package'].str.contains('1/2'), 'Price'] = price*2

new_pumpkins.head()
```

A basic scatterplot reminds us that we only have month data from August through December. We probably need more data to be able to draw conclusions in a linear fashion.So, we do DayofYear vs Price instead of month vs Price

```python
import matplotlib.pyplot as plt
plt.scatter('Month','Price',data=new_pumpkins)
```

```python
#To find correlation we use corr function.
print(new_pumpkins['Month'].corr(new_pumpkins['Price']))
print(new_pumpkins['DayOfYear'].corr(new_pumpkins['Price']))
```

Because the Price vs DayofYear scatter plot has several different clusters of points so there might be clusters of price vs pumpkin variety.

```python
colors=['red','blue','green','yellow']
ax=None
for i,var in enumerate(new_pumpkins['Variety'].unique()):
    df = new_pumpkins[new_pumpkins['Variety']==var]
    ax = df.plot.scatter('DayOfYear','Price',ax=ax,c=colors[i],label=var)
```

```python
#We drew a bar chart to confirm which variety is the cheapest.
new_pumpkins.groupby(['Variety'])['Price'].mean().plot(kind='bar')
#We grouped records on the basis of variety then got their price
```

```python
new_pumpkins = new_pumpkins[new_pumpkins['Variety'].str.contains('PIE TYPE', case=True, regex=True)]
print(new_pumpkins['DayOfYear'].corr(new_pumpkins['Price']))
```

```python
new_pumpkins.dropna(inplace=True)#Removes missing values aka. NaN
#Instead of removing those NaN values you can fill them with average value of that column
new_pumpkins.info()
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
from sklearn.model_selection import train_test_split

X=new_pumpkins['DayOfYear']
y=new_pumpkins['Price']
print(X.shape)
print(y.shape)
X=X.to_numpy().reshape((-1,1))#Because Linear Regression package expects 2D-array as input
print(X.shape)
```

```python
X_train, X_test, y_train, y_test=train_test_split(X, y, test_size=0.3, random_state=0)
#random_state=0 ensures that a data point is always assigned to the same set during each execution
```

```python
lin_regression=LinearRegression()
lin_regression.fit(X_train,y_train)#Training the model on training data
```

```python
print(lin_regression.coef_)#the coeff means the slope
print(lin_regression.intercept_)#Y intercept of the linear regression plot
```

```python
#To test the accuracy of our model we make prediction then find mse
Prediction=lin_regression.predict(X_test)
MeanSquareError=np.sqrt(mean_squared_error(y_test, Prediction))
print(f"The mean squared error: {MeanSquareError} ({MeanSquareError/np.mean(Prediction) * 100: 3.3}%)")
```

```python
#Coeff of determination is another metric for measuring accuracy
determination_coeff=lin_regression.score(X_train,y_train)
print("Coeff of determination: ",determination_coeff)
#Value of 0 means input data not taken into consideration. worst linear predictor
#Value of 1 means perfectly predicts all expected values.
```

#Polynomial Regression
We create a data pipeline(chain of estimators) using sklearns make_pipeline

```python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.pipeline import make_pipeline
```

```python
pipeline=make_pipeline(PolynomialFeatures(2), LinearRegression())
#PolynomialFeatures(2) means all the possible combinations of input data upto degree 2.
pipeline.fit(X_train,y_train)
```

```python
Polynomial_pred=pipeline.predict(X_test)
plt.scatter(X_test, y_test)
plt.plot(X_test, Polynomial_pred)
```

```python
X_test_sorted=X_test[np.argsort(X_test.ravel())]#X_test_sorted is 2D
#X_test.ravel() is 1D
Polynomial_pred=pipeline.predict(X_test_sorted)
plt.scatter(X_test, y_test)
plt.plot(X_test_sorted.ravel(), Polynomial_pred)
```

```python
New_Pumpkins = pd.DataFrame(
    {'Month': month, 
     'DayOfYear' : day_of_year, 
     'Variety': pumpkins['Variety'], 
     'City': pumpkins['City Name'], 
     'Package': pumpkins['Package'], 
     'Low Price': pumpkins['Low Price'],
     'High Price': pumpkins['High Price'], 
     'Price': price})
New_Pumpkins.loc[New_Pumpkins['Package'].str.contains('1 1/9'), 'Price'] = price/1.1
New_Pumpkins.loc[New_Pumpkins['Package'].str.contains('1/2'), 'Price'] = price*2

New_Pumpkins.head()
```

```python
pd.get_dummies(New_Pumpkins['Variety'])#Performs One-hot encoding on Variety attribute
```

```python
# set up training data
X = pd.get_dummies(New_Pumpkins['Variety']) \
        .join(New_Pumpkins['Month']) \
        .join(pd.get_dummies(New_Pumpkins['City'])) \
        .join(pd.get_dummies(New_Pumpkins['Package']))
y = New_Pumpkins['Price']

# make train-test split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)

# setup and train the pipeline
pipeline = make_pipeline(PolynomialFeatures(2), LinearRegression())
pipeline.fit(X_train,y_train)

# predict results for test data
pred = pipeline.predict(X_test)

# calculate MSE and determination
mse = np.sqrt(mean_squared_error(y_test,pred))
print(f'Mean error: {mse:3.3} ({mse/np.mean(pred)*100:3.3}%)')

score = pipeline.score(X_train,y_train)
print('Model determination: ', score)
```
