```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn import datasets, linear_model, model_selection
```

```python
import pandas as pd

X, y = datasets.load_linnerud(return_X_y=True, as_frame=True)
#X is the Exercise dataframe(i.e. X axis) while y is the physiological response dataframe (i.e. Y axis)
print(X.shape)
X = X.iloc[:, 1:2]
y = y.iloc[:, 1:2]
print(X.shape)
print(y.shape)
```

```python
D_train, D_test, T_train, T_test = model_selection.train_test_split(X, y, test_size=0.3)
Training_Model = linear_model.LinearRegression()
Training_Model.fit(D_train, T_train)#Training on train section of Data & Target
T_pred = Training_Model.predict(D_test)
```

```python
plt.scatter(D_test, T_test, color='black')
plt.plot(D_test, T_pred, color='red', linestyle='solid', marker='o')
plt.xlabel('Situps')
plt.ylabel('Waistline')
plt.title('Waistline vs Situp relationship prediction')
plt.show()
```
