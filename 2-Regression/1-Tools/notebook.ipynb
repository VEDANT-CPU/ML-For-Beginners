```python
from sklearn import datasets, linear_model, model_selection
```

```python
DataMat, Tar = datasets.load_diabetes(return_X_y=True)
print(DataMat.shape)
print(DataMat[0])
```

```python
DataMat = DataMat[:,2]
DataMat = DataMat.reshape((-1,1))
```

```python
#Splitting data and target into train and test set.
D_train, D_test, T_train, T_test = model_selection.train_test_split(DataMat, Tar, test_size=0.3)
#Now, train the model using linear_model module of sklearn
Mymodel = linear_model.LinearRegression()#No parameters needed
#fitting the model
Mymodel.fit(D_train,T_train)
```

```python
T_pred = Mymodel.predict(D_test)#Making predictions about test data
#then store those prediction in T_pred.
```

```python
plt.scatter(D_test, T_test,  color='black')
plt.plot(D_test, T_pred, color='blue', linewidth=3)
plt.xlabel('Scaled BMIs')
plt.ylabel('Disease Progression')
plt.title('A Graph Plot Showing Diabetes Progression Against BMI')
plt.show()
```
